## Iteration

Iteration is the process of repeatedly executing a block of code normally over a sequence (like a list).  In Python, this is typically done using `for` loops or `while` loops. On this session we are going to focus on `for` loops given that is the most commonly used in Python and in Data Engineering. 

__Why is this important in Data Engineering?__

1.  **Data Processing:** Iterating over data is fundamental to data engineering.  You'll use loops to process rows in a DataFrame, iterate over features, and cleand data.

2.  **Exploring Data:** You'll often use loops to explore data quality, visualize your data, calculating summary statistics or generating plots.

### __Example: Squaring numbers__

For this example, we would like to get the square of each item in the `nums` list. Without iteration, it would be like the following:

````python
# Example: Without Iteration (Manual Repetition)
nums = [1, 2, 3, 4, 5]
````

This is inefficient and inflexible. Let's use iteration.

````python
# Using iteration:
nums = [1, 2, 3, 4, 5]
````

### List Comprehensions in Python

List comprehensions are a concise, readable way to create lists in Python. They allow you to generate new lists by applying an expression to each item in an existing iterable, optionally filtering items based on a condition.

Think of them as a compact alternative to writing `for` loops that build up lists. They're not just shorter—they're often faster and more "Pythonic."


#### Basic Syntax

```python
new_list = [expression for item in iterable]
```

Breaking this down:
- **expression**: What you want to do with each item (transform it, keep it as-is, etc.)
- **item**: The variable representing each element as you iterate
- **iterable**: The source collection (list, range, string, etc.)

## Modular Coding

Modular coding means breaking down code into smaller, reusable pieces, usually as functions or modules.

__Why is this important in Data Engineering?__
 - Clean separation of **data preparation**, **feature engineering**, and **data loading**
 - Facilitates collaboration, testing, and reproducibility

- `def` keyword: Used to define a function.

### __Functions Exercise - Outlier detection__

Outlier detection and removal is crucial in data analysis because:

- __Statistical accuracy:__ Outliers can significantly skew statistical measures like means, variances, and correlations, leading to misrepresentation of the underlying data distribution.

- __Model performance:__ Machine learning models are highly sensitive to outliers, which can introduce bias, reduce predictive accuracy, and create unstable models that don't generalize well to new data.

- __Data quality assurance:__ Identifying outliers helps detect measurement errors, data entry mistakes, or equipment malfunctions that might otherwise go unnoticed.

- __Business insights:__ Removing noise from data enables clearer visualization and more reliable insights for decision-making.

- __Resource efficiency:__ Eliminating outliers can improve computational efficiency by reducing the complexity needed to fit models to unusual data points.
