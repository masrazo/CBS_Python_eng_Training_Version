**2XX Codes**<br>
<div class="alert alert-block alert-warning">
   These codes convey that the server has received the client’s request and processed it successfully. Here are the most common examples:
    <br>
</div>

**200 OK**: The request was successful.<br>
**201 Created**: The resource has been created on the server. This response is typically returned for POST requests.<br>
**202 Accepted**: The request has been received but is still being processed.<br>
**204 No Content**: The request has been successfully processed but no content will be returned.

```python
## Your code here: 
def movie_function(movie: str):
    """
        Funcition to get some movie facts.
        Parameters:
            movie (str): name of your movie
    """
    movies_api_key = 'd998895c'
    api_key = movies_api_key
    url = 'http://www.omdbapi.com/?t=' # ?t= refers to the title of the movie
    api_key_str = "&apikey=" + api_key
    response = requests.get(url + movie + api_key_str)
    data = response.json()
    director = data['Director']
    box_office = data['BoxOffice']
    ratings = data['Ratings']
    print("Movie Title:", movie)
    print("Director:", director)
    print("Box Office:", box_office)
    for source in range(len(ratings)):
        print(ratings[source]['Source'], ": ", ratings[source]['Value'])
 
movie_function('Home Alone')
#movie_function('Up')
 
```

**4XX Codes**<br>
<div class="alert alert-block alert-warning">
   These codes convey that the client’s request has an error. It’s possible that the request is written incorrectly, or the resource which the client is requesting doesn't exist.
    <br>
</div>

**400 Bad Request**: There is something wrong in the client’s request.<br>
**401 Unauthorized**: The client is not authorized to make this request.<br>
**403 Forbidden**: The request is correct but can’t be processed. Likely, the problem is that the client does not have required permissions.<br>
**404 Not Found**: The resource being requested doesn't exist.

```python
##Your code here:

##Your code here:

# Create an empty list to append our responses
multiresponse_json=[]

# Create a loop that will give back the response for all pages, the response 
counter = 1
while True: 
    query2= f"{url3}countries?&page={counter}&format=json"
    # Print the different pages for the query
    print(query2)
    response = requests.get(query2).json()
    if len(response[1]) != 0: 
        for county in response[1]:
            multiresponse_json.append(county)
        #multiresponse = response[1]
        counter += 1
    else:
        break

# Output
print("Items extracted: ", len(multiresponse_json))
print('Data sample:')
#pprint(multiresponse_json[0:3])
```
